```python
# Variable declarations
int_var = 10
float_var = 20.25
string_var = 'This is string varibable'
list_var = [1, 2, 30.33, 'This is another string', [20, 30, 40]]
bool_var = True
```

```python
# Printing to console
print(int_var)
print(float_var)
print(string_var)
print(list_var)
print(bool_var)
```

```python
# Conditional statements
if int_var <= 18:
    print("Adult")
else:
    print("Minor")
```

```python
# While loop
while int_var <= 15:
    print(int_var)
    int_var = int_var + 1
```

```python
# For loop
for var in range(0, 10):
    print(var)
```

```python
# Nested loops
for x in ["red", "big"]:
    for y in ["apple", "banana"]:
        print(x, y)
```
